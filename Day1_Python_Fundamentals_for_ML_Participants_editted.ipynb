## Loops — the `for` loop

A **loop** repeats a block of code. In ML, `for` loops show up constantly for **iterating through datasets**
(processing one row at a time) and for **training epochs** (repeating the training step a fixed number of times).

````python
# Demo: simulating training epochs with a for loop
for epoch in range(1, 6):
    loss = 1 / epoch   # a stand-in for a decreasing loss value
    print(f"Epoch {epoch}: loss = {loss:.3f}")
````

## Loops — the `while` loop

A `while` loop repeats **as long as a condition stays true** — useful when you don't know in advance how many
iterations you'll need. A classic ML example is **early stopping**: keep training until the loss drops below a
target value, however many epochs that takes.

````python
# Demo: training until the loss is small enough (early stopping)
loss = 1.0
epoch = 0

while loss > 0.1:
    loss = loss * 0.7   # loss shrinks by 30% each epoch
    epoch += 1
    print(f"Epoch {epoch}: loss = {loss:.4f}")

print(f"Stopped after {epoch} epochs, final loss = {loss:.4f}")
````

# Functions

Functions matter enormously in ML code because the same operation — normalizing a value, computing a loss,
scoring a prediction — needs to be applied over and over, often to every row in a dataset. Writing it once as a
function keeps your code short, consistent, and far easier to fix if something's wrong.

## Basic function structure

A function is defined with `def`, a name, parameters in parentheses, and a colon. `return` sends a value back to
wherever the function was called.

````python
# Demo: a reusable normalization function
def normalize(value, min_val, max_val):
    return (value - min_val) / (max_val - min_val)

scaled = normalize(75, 0, 100)
print("Scaled value:", scaled)

# The real power: apply it to many values at once
raw_scores = [45, 75, 90, 60]
scaled_scores = [normalize(s, 0, 100) for s in raw_scores]   # a compact loop — more on this shape later in the course
print("Scaled scores:", scaled_scores)
````

A **lambda** is a short, anonymous function written in a single line: `lambda arguments: expression`.
They're most useful when you need a small, throwaway function to hand to another function — very common in data

## What is a library, and what is NumPy?

## Creating arrays: Python list vs. NumPy array

A NumPy array looks similar to a list at first glance, but behaves very differently for math — and that difference
is the entire reason NumPy exists.

````python
# Demo: the key difference between a list and a NumPy array
python_list = [1, 2, 3, 4]
numpy_array = np.array([1, 2, 3, 4])

print("List * 2:  ", python_list * 2)     # repeats the list — probably NOT what you want
print("Array * 2: ", numpy_array * 2)     # multiplies every element — element-wise math

print("List + 10 would raise an error — try uncommenting the line below:")
# print(python_list + 10)
print("Array + 10:", numpy_array + 10)
````

## Mathematical operations

NumPy gives you fast, built-in aggregate functions — exactly the statistics you need before feeding data into a model:
`.mean()`, `.sum()`, `.std()` (standard deviation), `.max()`, `.min()`.

A very common preprocessing step is **standardization** (turning raw values into z-scores):
subtract the mean, then divide by the standard deviation.

````python
# Demo: summary statistics and standardization
prices = np.array([150000, 210000, 340000, 190000, 275000])

print("Mean:", prices.mean())
print("Std:", prices.std())
print("Max:", prices.max())
print("Min:", prices.min())

standardized = (prices - prices.mean()) / prices.std()
print("Standardized (z-scores):", standardized)
````

## Matrix operations

2D NumPy arrays behave as proper mathematical **matrices**. Two operations matter enormously for ML:

- **Transpose** (`.T`) — flips rows and columns.
- **Matrix multiplication** (`@`, or `np.dot()`) — combines a **feature matrix** with a **weight vector** to produce
  predictions. This single operation is the mathematical core of linear regression and the layers of a neural network.

If `X` is your feature matrix (rows = samples, columns = features) and `w` is a vector of weights (one per feature),
then `X @ w` gives you one prediction per sample — this *is* linear regression's forward pass.

````python
# Demo: matrix multiplication as a mini linear regression
X = np.array([[1, 2],
              [3, 4],
              [5, 6]])   # 3 samples, 2 features
weights = np.array([0.5, 1.5])       # one weight per feature

predictions = X @ weights
print("Predictions:", predictions)

print("Transpose of X:")
print(X.T)
````

# Mini ML Workflow — Combined Exercise

This final exercise is a **combined check** — it deliberately pulls together everything from today: variables and
operators, data structures, control flow, functions, and NumPy. It mirrors the real shape of an early ML pipeline:
**represent data → engineer a feature → label/classify → summarize with statistics.**

Work through the three steps below in order — each one builds on the last.

Run this cell first — it's our small "raw dataset" of houses, provided as a list of dictionaries (one dictionary per house).

````python
house_dataset = [
    {"sqft": 850,  "price": 150000},
    {"sqft": 1200, "price": 210000},
    {"sqft": 1750, "price": 300000},
    {"sqft": 2000, "price": 340000},
    {"sqft": 1300, "price": 225000},
    {"sqft": 950,  "price": 170000},
]

print(f"Loaded {len(house_dataset)} houses.")
````

### Step 1 — Engineer a feature and a label (functions + loop + dict + conditional)

1. Write a function `price_per_sqft(sqft, price)` that returns `price / sqft`, rounded to 2 decimals.
2. Loop through `house_dataset`. For each house dictionary:
   - compute its price-per-sqft using your function and store it under a new key `"ppsf"`
   - classify it as `"Expensive"` if `price > 250000`, otherwise `"Affordable"`, and store that under a new key `"label"`

````python
# TODO: define the function
def price_per_sqft(sqft, price):
    return price / sqft

# TODO: loop through house_dataset and add "ppsf" and "label" to each house
for house in house_dataset:
    house["ppsf"] = price_per_sqft(house["sqft"], house["price"])
    if house["price"] > 250000:
        house["label"] = "Expensive"
    else:
        house["label"] = "Affordable"

house_dataset   # displays the updated list — check that both new keys appear
````

### Step 2 — Dataset statistics with NumPy (lists → array + math)

1. Build a plain Python list `sqft_values` containing just the `sqft` from every house (loop through `house_dataset`).
2. Convert it to a NumPy array `sqft_array`.
3. Compute and print its `mean` and `std`.

````python
# TODO: build the plain list
sqft_values = []
for house in house_dataset:
    sqft_values.append(house["sqft"])

# TODO: convert to a NumPy array
sqft_array = np.array(sqft_values)

print("Mean sqft:", sqft_array.mean())
print("Std sqft:", sqft_array.std())
````

### Step 3 — Final summary report (loop + conditional counting + f-strings)

1. Loop through `house_dataset` again and count how many houses are labeled `"Expensive"`.
2. Print a short report: total number of houses, how many are Expensive vs. Affordable, and the mean/std of `sqft` from Step 2.

````python
# TODO: count expensive houses
expensive_count = 0
for house in house_dataset:
    if house["label"] == "Expensive":
        expensive_count += 1

affordable_count = len(house_dataset) - expensive_count

# TODO: print the final report using f-strings
print(f"Total houses: {len(house_dataset)}")
print(f"Expensive: {expensive_count}, Affordable: {affordable_count}")
print(f"Average size: {sqft_array.mean():.1f} sqft (std: {sqft_array.std():.1f})")
````
